```python
import pandas as pd
```

```python
# 1. Create a Pandas DataFrame with 3 columns and 5 rows filled with random integers.
#  Add a new column that is the product of the first two columns.
data={'X':[1,2,3,4,5],
     'Y':[4,3,5,8,10],
     'Z':[50,24,40,67,45]
}
df=pd.DataFrame(data)
df['X*Y']=(df.iloc[:,0])*(df.iloc[:,1])
df
```

```python
# 2. Create a Pandas DataFrame with 3 columns and 4 rows filled with random integers.
#  Compute the row-wise and column-wise sum.

data={'X':[23,45,56,69],
     'Y':[24,35,45,55],
     'Z':[5,10,15,20]
}
df=pd.DataFrame(data)
print(df)
row_sum=df.sum(axis=1)
column_sum=df.sum(axis=0)
print("\nRow Sum:")
print(row_sum)
print("\nColumn Sum:")
print(column_sum)
```

```python
# 3. Create a Pandas DataFrame with 3 columns and 5 rows filled with random integers.
# Introduce some NaN values.Fill the NaN values with the mean of the respective columns.

data={'X':[None,45,34,None,45],
     'Y':[None,30,None,34,67],
     'Z':[None,10,30,20,None]
}
df=pd.DataFrame(data)
df['Filled_X']=df['X'].fillna(df['X'].mean()).astype(int)
df['Filled_Y']=df['Y'].fillna(df['Y'].mean()).astype(int)
df['Filled_Z']=df['Z'].fillna(df['Z'].mean()).astype(int)
df
```

```python
# 4. Create a Pandas DataFrame with 4 columns and 6 rows filled with random integers.
#  Introduce some NaN values. Drop the rows with any NaN values.

data={'W':[80,45,34,56,45,67],
     'X':[None,30,34,34,67,50],
     'Y':[None,10,None,20,None,78],
     'Z':[34,67,89,90,23,56]
}
df=pd.DataFrame(data)
dropped=df.dropna()
dropped
```

```python
# 5. Data Aggregation 
# Create a Pandas DataFrame with 2 columns: 'Category' and 'Value'. 
# Fill the 'Category' column with random categories ('A', 'B', 'C')
# and the 'Value' column with random integers. Group the DataFrame by 'Category' 
# and compute the sum and mean of 'Value' for each category.

data={'Category': ['A','B','C','A','B','C'],
    'Value':[10,20,30,40,50,60]
}

df=pd.DataFrame(data)
result=df.groupby('Category')['Value'].agg(['sum','mean'])
print(result)
```

```python
# 6. Create a Pandas DataFrame with 3 columns: 'Product', 'Category', and 'Sales'.
# Fill the DataFrame with random data. Group the DataFrame by 'Category' 
# and compute the total sales for each category.
import random
data={
    'Product': ['Laptop', 'Mobile', 'Tablet', 'Mouse', 'Keyboard', 'Monitor'],
    'Category': ['A', 'B', 'C', 'A', 'B', 'C'],
    'Sales': [random.randint(100, 1000) for i in range(6)]
}

df = pd.DataFrame(data)
print("Original DataFrame:")
print(df)
result = df.groupby('Category')['Sales'].sum()
print("\nTotal Sales for Each Category:")
print(result)
```

```python
# 7. Merging DataFrames
# Create two Pandas DataFrames with a common column. Merge the DataFrames using the common column.

data1={'W':[80,45,34,56,45,67],
     'X':[29,30,34,34,67,50],
     'Y':[77,10,99,20,45,78],
     'Z':[34,67,89,90,23,56]
}
df1=pd.DataFrame(data1)

data2={'X':[29,30,34,34,67,50],
       'U':[1,2,3,4,5,6],
       'V':[7,8,9,10,11,12]
}
df2=pd.DataFrame(data2)
merged=pd.merge(df1,df2,on='X')
merged
```

```python
# 8. MultiIndex DataFrame
# Create a Pandas DataFrame with a MultiIndex (hierarchical index). 
# Perform some basic indexing and slicing operations on the MultiIndex DataFrame.
index = pd.MultiIndex.from_tuples([
        ('NY',2023),('NY',2024),
        ('LA',2023),('LA',2024)],names=['city','year'])

df=pd.DataFrame({'sales':[100, 120, 90, 95]},index=index)
print(df)
print(df.loc['NY'])
print(df.loc[('LA', 2024)])
print(df.sort_index().loc['LA':'NY'])
```

```python
# 9. Create a Pandas DataFrame with MultiIndex consisting of 'Category' and 'SubCategory'.
#  Fill the DataFrame with random data and compute the sum of values for each 'Category' and 'SubCategory'.

index=pd.MultiIndex.from_tuples(
    [('Electronics','A'),('Electronics','B'),
     ('Clothing','A'),('Clothing','B')],
    names=['Category','SubCategory']
)

df=pd.DataFrame({'Value':[50,30,20,40]},index=index)
print(df.groupby(level='Category').sum())
print(df.groupby(level='SubCategory').sum())
```

```python
# 10. Applying Functions
# Create a Pandas DataFrame with 3 columns and 5 rows filled with random integers.
#  Apply a function that doubles the values of the DataFrame.

df=pd.DataFrame({'A':[1,2,3],'B':[4,5,6],'C':[7,8,9]})
double=df.apply(lambda x: x * 2)
print(double)
```

```python
# 11. Create a Pandas DataFrame with 3 columns and 6 rows filled with random integers.
#  Apply a lambda function to create a new column that is the sum of the existing columns.

df=pd.DataFrame({'A':[1,2,3],'B':[4,5,6],'C':[7,8,9]})
df['Sum']=df.apply(lambda row:row['A']+row['B']+row['C'],axis=1)
print(df)
```
